### Demonstrates how to eliminate stopwords from a text using nltk

```python
# Removing stopwords is a critical preprocessing step for NLP tasks.
# It enhances the clarity of data, reduces noise, and focuses on meaningful
# features, which is crucial for accurate analysis in real-world scenarios
# like customer feedback, product reviews, and social media sentiment analysis.
```

```python
# Tokenization:
# The input text is split into individual words using word_tokenize().

# Stopwords Removal:
# A stopword is a commonly words that don't contribute to meaning, such as
# articles ('the','a', 'an'),
# prepositions ('in', 'at', # 'on'),
# and conjunctions ('and','but', 'or').
# This reduces noise in text analysis


# We retrieve the list of English stopwords from nltk and remove
# any matching words from the tokenized list.

# Filtering:
# The result shows only meaningful words, with stopwords eliminated.
```

```python
# Step 1: Install and Import Required Packages
# In Google Colab, NLTK is usually already installed.
import nltk

# Download the language resources used by the tokenizer and stopword list.
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# Step 2: Python Code to Eliminate Stopwords
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Define the sample text
text = """Natural language processing allows computers to
understand and respond to human language in a meaningful way."""

# Split the text into tokens (words and punctuation)
words = word_tokenize(text)

# Get the list of English stopwords
stop_words = set(stopwords.words('english'))

# Remove tokens that are in the stopword list
filtered_words = [word for word in words if word.lower() not in stop_words]

# Display the results
print("Original Words:", words)
print("Filtered Words (Without Stopwords):", filtered_words)
```

# Case Study-1

```python
# Problem: Optimizing User Queries for a Chatbot

# You are working as a Data Scientist for a customer support team building
# an AI-powered chatbot.
# The chatbot takes user queries as input and matches them against
# a knowledge base to provide relevant answers.
# However, you notice that the chatbot often struggles
# with queries that contain many stopwords (e.g., "the," "is," "at," "in," etc.),
# leading to irrelevant or incorrect responses.

# To improve the chatbot’s performance, you need to preprocess the user queries
# by eliminating stopwords before matching them with the knowledge base.
# This will help the chatbot focus only on the essential keywords,
# ensuring better matching and faster response times.
```

```python
# Objective: Write a Python program to:

# Tokenize each query into individual words.
# Remove stopwords using the nltk library.
# Display the list of keywords after stopword removal.

# Bonus Task:
# Write a function that allows the chatbot to take user input,
# removes the stopwords, and returns the optimized query.

# Challenge:
# Implement a custom stopword list for words specific to your company,
# like "customer," "order," or "product," to further enhance the
# chatbot’s performance.
```

```python
# This problem simulates a real-world NLP challenge where stopword removal is essential for improving chatbot or search engine performance, ensuring more accurate keyword extraction and matching.
```

```python
# Conclusion
# This solution:
# Tokenizes the user queries into words.
# Removes stopwords using nltk to focus on essential keywords.
# Handles custom stopwords, improving the chatbot’s performance by
# excluding irrelevant business terms.
# Provides a chatbot input function to optimize queries on the fly.
# This approach ensures that your chatbot processes queries more effectively,
# leading to faster and more accurate responses.
```

```python
# Case Study-1: Optimizing User Queries for a Chatbot
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Load English stopwords
stop_words = set(stopwords.words('english'))

# Add company-specific stopwords requested in the task
custom_stopwords = {'customer', 'order', 'product'}
stop_words.update(custom_stopwords)

def optimize_query(query):
    """Return a query after removing English and custom stopwords."""
    words = word_tokenize(query)
    keywords = [
        word for word in words
        if word.isalnum() and word.lower() not in stop_words
    ]
    return keywords

# Example customer queries
queries = [
    "How can I track my order?",
    "Where is the customer support team?",
    "Can I check the product delivery status?"
]

for query in queries:
    print("Original query:", query)
    print("Keywords after stopword removal:", optimize_query(query))
    print()

# Bonus task: allow a user to enter their own query
user_query = input("Enter a customer query: ")
print("Optimized query:", optimize_query(user_query))
```

# Case Study-2

```python
# Objective: Analyze customer reviews to identify key topics and
# sentiments without noise from irrelevant words.

# Dataset:
# A dataset of customer reviews about a restaurant:

# "The food was absolutely fantastic and the service was great!"
# "The ambiance was okay, but the food was cold and not tasty."
# "I loved the experience. The staff was friendly and polite."
```

```python
# Step 1: Without Removing Stopwords
# Using raw text for sentiment analysis or
# topic modeling includes unnecessary words, leading to:
# Sentiment scores influenced by neutral words like "the" or "and."
# Topics heavily populated with stopwords.
```

```python
# Common stopwords dominate the word counts, diluting meaningful insights.
```

```python
# Step 2: Removing Stopwords
# By removing stopwords, we retain only the meaningful words,
# improving the performance of NLP tasks.
```

```python
# Keywords like "food," "service," "great," and "fantastic" emerge as meaningful insights.
```

```python
# Impact on Applications
# Sentiment Analysis:

# Sentiment scores become more accurate by focusing on emotional words
# like "fantastic" or "cold."
# Stopwords removal avoids neutralizing sentiment polarity.

# Text Summarization:
# Generated summaries focus on essential content
# (e.g., "food fantastic," "service great") without redundancy.

# Topic Modeling:
# Topics clearly identify themes like "food," "service," or "ambiance,"
# leading to better customer experience analysis.
```

```python
# Case Study-2: Removing Stopwords from Restaurant Reviews
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

reviews = [
    "The food was absolutely fantastic and the service was great!",
    "The ambiance was okay, but the food was cold and not tasty.",
    "I loved the experience. The staff was friendly and polite."
]

stop_words = set(stopwords.words('english'))

# Keep negation words such as 'not' because they change the review's meaning.
stop_words.discard('not')
stop_words.discard("n't")

for number, review in enumerate(reviews, start=1):
    words = word_tokenize(review)
    filtered_words = [
        word for word in words
        if word.isalnum() and word.lower() not in stop_words
    ]
    print(f"Review {number}")
    print("Original review:", review)
    print("Words after stopword removal:", filtered_words)
    print()
```
